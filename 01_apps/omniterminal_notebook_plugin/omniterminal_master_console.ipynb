```python
%%omni_polyglot python
# Practice: Implement non-blocking PTY ring buffer byte drainer
def drain_ring_buffer(capacity: int, chunk_size: int = 4096) -> dict:
    transferred = 0
    while transferred < capacity:
        transferred += min(chunk_size, capacity - transferred)
    return {'transferred_bytes': transferred, 'buffer_empty': True}

res = drain_ring_buffer(65536)
assert res['transferred_bytes'] == 65536
assert res['buffer_empty'] is True
print('RING_BUFFER_DRAIN_VERIFIED_OK')
```
