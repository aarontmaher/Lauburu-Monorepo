```python
import os
import sys
import json
import time
import sqlite3
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

REPO_ROOT = Path('/Users/aaron/DFS_UNIFIED/Lauburu-Monorepo')
LENS_DB = Path(os.path.expanduser('~/.lauburu/screen_lens.sqlite'))

print('👁️ Screen Lens Visual Cortex Connected.')
print(f'• Database Path: {LENS_DB} (Exists: {LENS_DB.exists()})')
```

## 📊 1. Real-Time Developer Activity & Focus Timeline

```python
if LENS_DB.exists():
    conn = sqlite3.connect(f'file:{LENS_DB}?mode=ro', uri=True)
    df_events = pd.read_sql_query("""
        SELECT id, timestamp, active_app_name, active_window_title, 
               capture_duration_ms, ocr_duration_ms, raw_text_summary
        FROM captures ORDER BY id DESC LIMIT 200
    """, conn)
    conn.close()
    
    print(f'✅ Fetched {len(df_events)} recent activity frames.')
    
    # Compute active app share
    app_share = df_events['active_app_name'].value_counts()
    
    fig, ax = plt.subplots(figsize=(10, 4))
    app_share.head(6).plot(kind='barh', ax=ax, color='#17becf', edgecolor='black')
    ax.set_title('Live Developer Focus by Application (Last 200 Activity Spikes)')
    ax.set_xlabel('Capture Frames')
    plt.tight_layout()
    plt.show()
else:
    print('⚠️ Screen Lens database not found.')
```

## 🗺️ 2. Dynamic Monorepo Subsystem Mapping (Where is Focus Flowing?)

```python
subsystem_keywords = {
    '00_core_infrastructure': ['docker', 'tailscale', 'seaweedfs', 'ports', 'systemd'],
    '01_apps': ['screen_lens', 'movesense_hub', 'zone2', 'spatial', 'flutter', 'notebooks'],
    '02_ai_models_and_inference': ['llama', 'prima', 'petals', 'exo', 'gguf', 'tui', 'weights'],
    '03_biometrics_and_telemetry': ['ecg', 'pan-tompkins', 'dfa', 'heart', 'movesense'],
    '04_data_and_memory': ['lora', 'pyspark', 'qdrant', 'ast', 'datasets', 'parquet'],
    '05_agents_and_swarms': ['debate', 'swarm', 'orchestrator', 'moe', 'devil', 'elo'],
    '06_scripts_and_tooling': ['router', 'ssh', 'adb', 'keepalive', 'wol', 'pytest'],
    '07_docs_and_architecture': ['rfc', 'whitepaper', 'obsidian', 'taxonomy', 'rules']
}

if LENS_DB.exists():
    subsystem_scores = {k: 0 for k in subsystem_keywords}
    for text in df_events['raw_text_summary'].dropna():
        t_lower = text.lower()
        for sub, kw_list in subsystem_keywords.items():
            if any(kw in t_lower for kw in kw_list):
                subsystem_scores[sub] += 1
                
    df_subs = pd.DataFrame(list(subsystem_scores.items()), columns=['Subsystem', 'Focus Score'])
    df_subs = df_subs.sort_values(by='Focus Score', ascending=False)
    
    print('🏛️ Dynamic Monorepo Activity Distribution:')
    display(df_subs)
    
    plt.figure(figsize=(10, 3.5))
    plt.bar(df_subs['Subsystem'], df_subs['Focus Score'], color='#9467bd', edgecolor='black')
    plt.xticks(rotation=25, ha='right')
    plt.title('Monorepo Subsystem Immersion Index (Derived from Live Screen Lens)')
    plt.ylabel('Activity Matches')
    plt.tight_layout()
    plt.show()
```

## 🌐 4. Chrome DevTools Protocol (CDP), Computer & Device Use Bridge
Connects to local and remote Chrome instances (Mac Host, Android Pixel 10 Pro XL, Samsung S20+) for automated visual inspection, DOM tree analysis, and real-time a11y auditing.


```python
# 4. Interactive Chrome DevTools & Multi-Device Inspector
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

BRIDGE_PATH = Path("/Users/aaron/DFS_UNIFIED/Lauburu-Monorepo/01_apps/lauburu_lens/chrome_debugging_bridge.py")

device_select = widgets.Dropdown(
    options=[
        ("🖥️ Local Mac Host (Port 9222)", "127.0.0.1:9222"),
        ("📱 Layer 6: Pixel 10 Pro XL (ADB tcp:9223)", "100.73.38.87:5555"),
        ("📱 Layer 7: Samsung S20+ (ADB tcp:9224)", "100.84.40.95:5555"),
        ("🐧 Layer 3: Linux Head Node (CDP tcp:9222)", "100.101.39.98:9222")
    ],
    value="127.0.0.1:9222",
    description="Target Node:",
    layout=widgets.Layout(width='420px')
)

target_url_input = widgets.Text(
    value="http://127.0.0.1:8890",
    description="Inspect URL:",
    layout=widgets.Layout(width='400px')
)

audit_btn = widgets.Button(description="🔍 Run Lens Chrome Audit", button_style='primary', icon='search')
adb_btn = widgets.Button(description="📱 Forward Android CDP via ADB", button_style='info', icon='mobile')
cdp_out = widgets.Output()

def run_lens_audit(b):
    with cdp_out:
        clear_output()
        print(f"[{time.strftime('%H:%M:%S')}] 🔍 Auditing UI at {target_url_input.value} on {device_select.label}...")
        try:
            res = subprocess.run([sys.executable, str(BRIDGE_PATH)], capture_output=True, text=True)
            print(res.stdout)
        except Exception as e:
            print(f"Audit error: {e}")

def run_adb_forward(b):
    with cdp_out:
        clear_output()
        dev_id = device_select.value
        print(f"[{time.strftime('%H:%M:%S')}] 📱 Forwarding Chrome DevTools on {dev_id}...")
        cmd = ["adb", "-s", dev_id, "forward", "tcp:9223", "localabstract:chrome_devtools_remote"]
        r = subprocess.run(cmd, capture_output=True, text=True)
        if r.returncode == 0:
            print(f"✅ Successfully forwarded to http://127.0.0.1:9223")
        else:
            print(f"⚠️ ADB notice: {r.stderr.strip() or 'Port forwarded / ready'}")

audit_btn.on_click(run_lens_audit)
adb_btn.on_click(run_adb_forward)

display(widgets.VBox([
    widgets.HBox([device_select, target_url_input]),
    widgets.HBox([audit_btn, adb_btn]),
    cdp_out
]))
```
