# 🫀 Movesense 512Hz ECG & Pan-Tompkins DSP Analysis
Applies 5-15 Hz bandpass filtering, derivative, squaring, and moving-window integration.

```python
import numpy as np

def pan_tompkins_filter_specs(fs=512.0):
    nyq = 0.5 * fs
    low, high = 5.0 / nyq, 15.0 / nyq
    return {'sampling_rate': fs, 'bandpass_norm': (low, high), 'qrs_window_ms': 150}

specs = pan_tompkins_filter_specs()
print(f'✅ DSP Filter Initialized: {specs}')
```
