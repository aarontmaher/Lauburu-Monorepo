```python
import json, urllib.request
from pathlib import Path

def test_debate_connectivity():
    ports = {'local_orchestrator': 8081, 'prima_cpp': 8082, 'devils_advocate': 8083}
    results = {}
    for name, port in ports.items():
        try:
            with urllib.request.urlopen(f'http://127.0.0.1:{port}/health', timeout=1.0) as resp:
                results[name] = resp.status == 200
        except Exception:
            results[name] = False
    return results

status = test_debate_connectivity()
print(f'✅ Tri-Orchestrator Port Health: {status}')
```
